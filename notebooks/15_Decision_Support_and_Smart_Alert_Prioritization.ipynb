```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("../data/final/hybrid_ai_risk_assessment.csv")

print("Dataset Loaded Successfully")
print()

print("Shape :", df.shape)

df.head()
```

```python

print("Dataset Information")


print()

print(df.info())

print()


print("Missing Values")


print(df.isnull().sum())
```

```python
required_columns = [

    "callsign",

    "Hybrid_Flight_Risk_Score",

    "Hybrid_Risk_Category",

    "Behavioural_Intelligence_Score",

    "Operational_Intelligence_Score",

    "GMM_Anomaly_Label",

    "ELM_Anomaly_Label"

]

missing = []

for col in required_columns:

    if col not in df.columns:

        missing.append(col)

if len(missing)==0:

    print("All required columns are available.")

else:

    print("Missing Columns")

    print(missing)
```

```python
df["Hybrid_Flight_Risk_Score"].describe()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Hybrid_Risk_Category",
    order=["Low","Medium","High","Critical"]
)

plt.title("Hybrid Risk Category Distribution")

plt.xlabel("Risk Category")

plt.ylabel("Number of Flights")

plt.show()
```

```python
def assign_alert_level(score):

    if score >= 80:
        return "Critical"

    elif score >= 60:
        return "High"

    elif score >= 30:
        return "Medium"

    else:
        return "Routine"


df["OCC_Alert_Level"] = df["Hybrid_Flight_Risk_Score"].apply(assign_alert_level)

print("Operational Alert Levels Assigned Successfully")

df["OCC_Alert_Level"].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="OCC_Alert_Level",
    order=["Routine","Medium","High","Critical"]
)

plt.title("Operational Alert Level Distribution")

plt.xlabel("Alert Level")

plt.ylabel("Number of Flights")

plt.show()
```

```python
def occ_recommendation(level):

    if level == "Critical":

        return "Immediate operational intervention required"

    elif level == "High":

        return "Notify OCC supervisor and increase monitoring"

    elif level == "Medium":

        return "Continue enhanced monitoring"

    else:

        return "Routine monitoring"


df["OCC_Recommendation"] = df["OCC_Alert_Level"].apply(occ_recommendation)

print("Recommendations Generated Successfully")

df[[
    "Hybrid_Flight_Risk_Score",
    "OCC_Alert_Level",
    "OCC_Recommendation"
]].head()
```

```python
df["Flight_Priority_Rank"] = (
    df["Hybrid_Flight_Risk_Score"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

print("Flight Priority Ranking Generated Successfully")

df[[
    "callsign",
    "Hybrid_Flight_Risk_Score",
    "Hybrid_Risk_Category",
    "OCC_Alert_Level",
    "Flight_Priority_Rank"
]].head()
```

```python
alert_summary = (
    df["OCC_Alert_Level"]
    .value_counts()
    .rename_axis("Alert Level")
    .reset_index(name="Number of Flights")
)

alert_summary["Percentage"] = (
    alert_summary["Number of Flights"] /
    len(df) * 100
).round(2)

alert_summary
```

```python
critical = len(df[df["OCC_Alert_Level"] == "Critical"])

high = len(df[df["OCC_Alert_Level"] == "High"])

medium = len(df[df["OCC_Alert_Level"] == "Medium"])

routine = len(df[df["OCC_Alert_Level"] == "Routine"])


print("Information Overload Analysis")


print(f"Total Flights                    : {len(df)}")
print(f"Routine Monitoring               : {routine}")
print(f"Medium Monitoring                : {medium}")
print(f"High Priority                    : {high}")
print(f"Critical Priority                : {critical}")
print()
print(f"Flights Requiring Immediate Attention (High + Critical): {high + critical}")
print(f"Percentage of Total Flights: {((high + critical)/len(df))*100:.2f}%")
```
